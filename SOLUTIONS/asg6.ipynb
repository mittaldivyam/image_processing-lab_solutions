```python
#q1
import cv2
import numpy as np

# Load image
img = cv2.imread('input4.jpg', 0)

# Take input
k = int(input("Enter kernel size (odd): "))

# Create averaging kernel
kernel = np.ones((k, k), np.float32) / (k * k)

# Apply filter
output = cv2.filter2D(img, -1, kernel)

# Show results
cv2.imshow("Original", img)
cv2.imshow("Averaging Filter", output)
cv2.waitKey(0)
cv2.destroyAllWindows()
 # Best choice: 3
```

```python
#q2 
import cv2
import numpy as np

img = cv2.imread('input4.jpg', 0)

k = int(input("Enter kernel size (odd): "))
center = k // 2

kernel = np.zeros((k, k), dtype=np.float32)

# Create weighted kernel
for i in range(k):
    for j in range(k):
        x = abs(i - center)
        y = abs(j - center)
        kernel[i, j] = max(x, y)

# Normalize kernel
kernel = kernel / np.sum(kernel)

# Apply filter
output = cv2.filter2D(img, -1, kernel)

cv2.imshow("Original", img)
cv2.imshow("Weighted Filter", output)
cv2.waitKey(0)
cv2.destroyAllWindows()

 # Best choice: 5
```

```python
#q3
import cv2

img = cv2.imread('input4.jpg', 0)

k = int(input("Enter kernel size (odd): "))
sigma = float(input("Enter sigma: "))

# Apply Gaussian filter
output = cv2.GaussianBlur(img, (k, k), sigma)

cv2.imshow("Original", img)
cv2.imshow("Gaussian Filter", output)
cv2.waitKey(0)
cv2.destroyAllWindows()

# Enter kernel size (odd): 5
# Enter sigma: 1.0
```
