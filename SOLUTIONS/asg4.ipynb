```python
#q1
import cv2
import numpy as np

# Read image
img = cv2.imread('input1.jpg', 0)  # grayscale

# Apply log transform
c = 1
log_transformed = c * np.log(1 + img)

# Normalize to 0–255
log_transformed = np.array(255 * log_transformed / np.max(log_transformed), dtype=np.uint8)

# Show output
cv2.imshow("Original", img)
cv2.imshow("Log Transform", log_transformed)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#q2
import cv2
import numpy as np

# Read image
img = cv2.imread('input2.jpg', 0)

# Take gamma input
gamma = float(input("Enter gamma value: "))

# Apply gamma correction
c = 1
gamma_corrected = c * np.power(img / 255.0, gamma)

# Scale back to 0–255
gamma_corrected = np.array(gamma_corrected * 255, dtype=np.uint8)

# Show output
cv2.imshow("Original", img)
cv2.imshow("Gamma Corrected", gamma_corrected)
cv2.waitKey(0)
cv2.destroyAllWindows()
# agar hum gamma ki value 1 se kam daalenge toh bright image warna dark image
```

```python
#q3
import cv2
import numpy as np

# Read image
img = cv2.imread('input3.jpg', 0)

# User input
r1 = int(input("Enter r1 (input min): "))
r2 = int(input("Enter r2 (input max): "))
s1 = int(input("Enter s1 (output min): "))
s2 = int(input("Enter s2 (output max): "))

# Apply contrast stretching
output = np.zeros_like(img)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        r = img[i, j]
        
        if r < r1:
            output[i, j] = 0
        elif r <= r2:
            output[i, j] = ((r - r1) * (s2 - s1) / (r2 - r1)) + s1
        else:
            output[i, j] = 255

# Convert to uint8
output = np.array(output, dtype=np.uint8)

# Show output
cv2.imshow("Original", img)
cv2.imshow("Contrast Stretched", output)
cv2.waitKey(0)
cv2.destroyAllWindows()
# 0 ≤ r1 < r2 ≤ 255
# 0 ≤ s1 < s2 ≤ 255
# C:\Users\NEW\AppData\Local\Temp\ipykernel_27936\924280072.py:24: RuntimeWarning: overflow encountered in scalar multiply
#   output[i, j] = ((r - r1) * (s2 - s1) / (r2 - r1)) + s1 in this error when we perform the formula given it gives values more than 255 that is why 
# this is showing an error 
```
