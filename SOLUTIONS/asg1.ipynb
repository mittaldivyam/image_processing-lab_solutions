```python
import cv2                      # OpenCV library for image processing
import numpy as np             # For numerical operations
from matplotlib import pyplot as plt   # For displaying images in Jupyter

# Step 1: Read the image in grayscale mode (0 means grayscale)
img = cv2.imread("input.jpg", 0)

# Step 2: Check if image is loaded properly
if img is None:
    print("Error: Image not found")
    exit()

# Step 3: Calculate mean intensity of the image
# This will act as threshold value
threshold = int(np.mean(img))
print("Mean Threshold Value:", threshold)

# Step 4: Create an empty binary image of same size
binary = np.zeros_like(img)

# Step 5: Apply thresholding manually (pixel by pixel)
for i in range(img.shape[0]):          # Loop over rows
    for j in range(img.shape[1]):      # Loop over columns
        
        # If pixel value is greater than or equal to threshold
        if img[i, j] >= threshold:
            binary[i, j] = 255         # Set to white
        else:
            binary[i, j] = 0           # Set to black

# Step 6: Display the result
plt.imshow(binary, cmap='gray')       # Show as grayscale image
plt.title("Binary Image (Mean Threshold)")
plt.axis('off')                       # Hide axis
plt.show()
```

```python
import cv2                      # OpenCV for image processing
import numpy as np             # Numerical operations
from matplotlib import pyplot as plt   # Display image in Jupyter

# Step 1: Read the image in grayscale mode
img = cv2.imread("input.jpg", 0)

# Step 2: Check if image is loaded properly
if img is None:
    print("Error: Image not found. Make sure input.jpg is in the same folder.")
    exit()

# Step 3: Take threshold value from user
threshold = int(input("Enter threshold value (0-255): "))

# Step 4: Create empty binary image of same size
binary = np.zeros_like(img)

# Step 5: Apply thresholding pixel by pixel
for i in range(img.shape[0]):        # Loop over rows
    for j in range(img.shape[1]):    # Loop over columns
        
        if img[i, j] >= threshold:
            binary[i, j] = 255       # White pixel
        else:
            binary[i, j] = 0         # Black pixel

# Step 6: Display the result
plt.imshow(binary, cmap='gray')
plt.title("Binary Image (User Threshold)")
plt.axis('off')
plt.show()
```
