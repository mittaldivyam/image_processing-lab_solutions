```python
import cv2
import numpy as np

# Read image
img = cv2.imread("input.jpg", cv2.IMREAD_GRAYSCALE)  
# use IMREAD_COLOR if you want color image

# Input from user
pad = int(input("Enter border width: "))
color = int(input("Enter border color (0-255): "))

# Add border
padded = cv2.copyMakeBorder(
    img,
    pad, pad, pad, pad,
    cv2.BORDER_CONSTANT,
    value=color
)

# Show result
cv2.imshow("Original", img)
cv2.imshow("Padded Image", padded)

cv2.waitKey(0)
cv2.destroyAllWindows()

# Save output
cv2.imwrite("padded_output.jpg", padded)
```

```python
import cv2
import numpy as np

# Read image
img = cv2.imread("input.jpg", cv2.IMREAD_GRAYSCALE)

# Ask type
type_img = int(input("Enter type (1 = Binary, 2 = Grayscale): "))

if type_img == 1:
    # Binary complement
    _, img = cv2.threshold(img, 127, 1, cv2.THRESH_BINARY)
    complement = 1 - img

    # Convert back to visible format (0-255)
    complement = complement * 255

else:
    # Grayscale complement
    complement = 255 - img

# Show result
cv2.imshow("Original", img)
cv2.imshow("Complement", complement)

cv2.waitKey(0)
cv2.destroyAllWindows()

# Save output
cv2.imwrite("complement_output.jpg", complement)
```
