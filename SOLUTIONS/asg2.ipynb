```python
import cv2                      # OpenCV for image processing
import numpy as np             # Numerical operations
from matplotlib import pyplot as plt   # Display images

# Step 1: Read the color image (default = BGR format)
img = cv2.imread("input.jpg")

# Step 2: Check if image is loaded
if img is None:
    print("Error: Image not found")
    exit()

# Step 3: Get image dimensions
h, w, _ = img.shape

# Step 4: Create empty grayscale image
gray = np.zeros((h, w), dtype=np.uint8)

# Step 5: Convert RGB to grayscale using mean method
for i in range(h):
    for j in range(w):
        
        B, G, R = img[i, j]   # OpenCV uses BGR
        
        # Mean formula
        gray_value = int((R + G + B) / 3)
        
        gray[i, j] = gray_value

# Step 6: Display result
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image (Mean Method)")
plt.axis('off')
plt.show()
```

```python
import cv2                      # OpenCV for image processing
import numpy as np             # Numerical operations
from matplotlib import pyplot as plt   # Display images

# Step 1: Read the image (color image)
img = cv2.imread("input.jpg")

# Step 2: Check if image is loaded
if img is None:
    print("Error: Image not found. Make sure input.jpg is in the same folder.")
    exit()

# Convert image to higher datatype to avoid overflow
img = img.astype("int32")

# Step 3: Get dimensions
h, w, _ = img.shape

# Step 4: Create empty grayscale images
gray_mean = np.zeros((h, w), dtype=np.uint8)
gray_weighted = np.zeros((h, w), dtype=np.uint8)

# =========================
# 🔹 CASE 1: MEAN METHOD
# =========================
for i in range(h):
    for j in range(w):
        
        B, G, R = img[i, j]   # OpenCV uses BGR
        
        gray_value = (R + G + B) // 3
        
        gray_mean[i, j] = gray_value

# =========================
# 🔹 CASE 2: WEIGHTED METHOD
# =========================

# Take weights from user
wr = float(input("Enter weight for R: "))
wg = float(input("Enter weight for G: "))
wb = float(input("Enter weight for B: "))

# Check sum of weights
if abs((wr + wg + wb) - 1.0) > 1e-6:
    print("Error: Weights must sum to 1")
    exit()

for i in range(h):
    for j in range(w):
        
        B, G, R = img[i, j]
        
        gray_value = int(wr * R + wg * G + wb * B)
        
        gray_weighted[i, j] = gray_value

# =========================
# DISPLAY RESULTS
# =========================

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(gray_mean, cmap='gray')
plt.title("Grayscale (Mean Method)")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(gray_weighted, cmap='gray')
plt.title("Grayscale (Weighted Method)")
plt.axis('off')

plt.show()
```
